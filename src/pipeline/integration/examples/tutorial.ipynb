## Pipeline


A Pipeline is a list of steps, executed sequentially one after the other. Similar to a step, you may instantiate a pipeline:

- With direct instantiation
- With a dictionary
- With a YAML config

When instantiating direcly or with a dictionary, you may also instantiate the steps directly or with a dictionary, or with a mix of both. To instantiate a step, use the following format:

```python
{
    "step_type": "", #The class name of the Step (e.g. HeuristicFilter, MinHashDeduplication)}
    "args":{}        #The arguments needed to instantiate the Step
}
```

````python
from src.pipeline.integration import Pipeline
````

### List of Steps sample

````python
pipeline_steps = [
    URLDeduplication(
        duplicate_ids_path="outputs/deduplication/url_202606011252/duplicate_ids.json",
        ignore_files_path="outputs/deduplication/url_dedup_ignore_corpora.json"
    ),
    {
        "step_type":"MinHashDeduplication",
        "args": {
            "duplicate_ids_path":"outputs/deduplication/minhash_202606111735/duplicates.json"
        }
    }
]
````

### Direct instantiation

````python
pipeline_direct = Pipeline(
    steps=pipeline_steps,
    input_directory="data/processed",
    output_directory="outputs/sample_outputs/pipeline_direct",
    start=1
)
````

````python
pipeline_direct.run()
````

### From a dictionary

````python
pipeline_dict_config = {
    "steps": pipeline_steps,
    "input_directory":"data/processed",
    "output_directory":"outputs/sample_outputs/pipeline_direct",
    "start":1
}

pipeline_dict = Pipeline.from_dict(pipeline_dict_config)
````

````python
pipeline_dict.run()
````

### From a YAML file

````python
pipeline_yaml = Pipeline.from_yaml(yaml_path="src/pipeline/integration/examples/configs/pipeline.yaml")
````

````python
pipeline_yaml.run()
````
